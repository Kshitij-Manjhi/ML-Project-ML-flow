Cell 11 — Extract MFCC features from one utterance

```python
import numpy as np
import librosa

# Extract 13 MFCC coefficients
mfcc = librosa.feature.mfcc(
    y=y,
    sr=sr,
    n_mfcc=13
)

print("MFCC shape:", mfcc.shape)

# Convert each MFCC coefficient into mean + standard deviation
mfcc_mean = np.mean(mfcc, axis=1)
mfcc_std = np.std(mfcc, axis=1)

print("\nMFCC means:")
print(mfcc_mean)

print("\nMFCC standard deviations:")
print(mfcc_std)
```

Cell 12 — Extract a complete acoustic feature vector

```python
# Spectral features
spectral_centroid = librosa.feature.spectral_centroid(y=y, sr=sr)
spectral_bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr)
spectral_rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr)

# Zero crossing rate
zcr = librosa.feature.zero_crossing_rate(y)

# RMS energy
rms = librosa.feature.rms(y=y)

# Combine everything into one feature dictionary
features = {}

# MFCC mean + std
for i in range(13):
    features[f"mfcc_{i+1}_mean"] = np.mean(mfcc[i])
    features[f"mfcc_{i+1}_std"] = np.std(mfcc[i])

# Other acoustic features
features["spectral_centroid_mean"] = np.mean(spectral_centroid)
features["spectral_centroid_std"] = np.std(spectral_centroid)

features["spectral_bandwidth_mean"] = np.mean(spectral_bandwidth)
features["spectral_bandwidth_std"] = np.std(spectral_bandwidth)

features["spectral_rolloff_mean"] = np.mean(spectral_rolloff)
features["spectral_rolloff_std"] = np.std(spectral_rolloff)

features["zcr_mean"] = np.mean(zcr)
features["zcr_std"] = np.std(zcr)

features["rms_mean"] = np.mean(rms)
features["rms_std"] = np.std(rms)

# Duration
features["duration"] = len(y) / sr

print("Number of features:", len(features))
print("\nFeatures:")
print(features)
```

Cell 13 — Create the feature extraction function

```python
def extract_features(y, sr):
    features = {}

    # MFCC
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)

    for i in range(13):
        features[f"mfcc_{i+1}_mean"] = np.mean(mfcc[i])
        features[f"mfcc_{i+1}_std"] = np.std(mfcc[i])

    # Spectral features
    spectral_centroid = librosa.feature.spectral_centroid(y=y, sr=sr)
    spectral_bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr)
    spectral_rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr)

    # Zero crossing rate
    zcr = librosa.feature.zero_crossing_rate(y)

    # RMS energy
    rms = librosa.feature.rms(y=y)

    features["spectral_centroid_mean"] = np.mean(spectral_centroid)
    features["spectral_centroid_std"] = np.std(spectral_centroid)

    features["spectral_bandwidth_mean"] = np.mean(spectral_bandwidth)
    features["spectral_bandwidth_std"] = np.std(spectral_bandwidth)

    features["spectral_rolloff_mean"] = np.mean(spectral_rolloff)
    features["spectral_rolloff_std"] = np.std(spectral_rolloff)

    features["zcr_mean"] = np.mean(zcr)
    features["zcr_std"] = np.std(zcr)

    features["rms_mean"] = np.mean(rms)
    features["rms_std"] = np.std(rms)

    # Duration
    features["duration"] = len(y) / sr

    return features


# Test the function on our current audio
test_features = extract_features(y, sr)

print("Number of features:", len(test_features))
print("Feature names:")
print(list(test_features.keys()))
```

Cell 14 — Select 1,000 utterances

Cell 15 — Extract features from 1,000 audio files

```python
import time
import io
import tarfile
import pandas as pd

sample_ids = set(sample_metadata["Uttids"])

feature_rows = []
start_time = time.time()

with tarfile.open(dataset_path, "r:gz") as tar:
    for i, member in enumerate(tar):

        # Only process MP3 files
        if not member.isfile() or not member.name.endswith(".mp3"):
            continue

        uttid = member.name.split("/")[-1].replace(".mp3", "")

        # Only process our selected 1,000 utterances
        if uttid not in sample_ids:
            continue

        try:
            audio_bytes = tar.extractfile(member).read()

            y, sr = librosa.load(
                io.BytesIO(audio_bytes),
                sr=16000,
                mono=True
            )

            features = extract_features(y, sr)
            features["Uttids"] = uttid

            feature_rows.append(features)

            if len(feature_rows) % 100 == 0:
                print(
                    f"Processed {len(feature_rows)}/1000 | "
                    f"Time: {(time.time()-start_time)/60:.1f} min"
                )

            if len(feature_rows) == 1000:
                break

        except Exception as e:
            print(f"Error processing {uttid}: {e}")

# Convert to DataFrame
features_df = pd.DataFrame(feature_rows)

print("\nDone!")
print("Feature matrix shape:", features_df.shape)
print("Time taken:", round((time.time()-start_time)/60, 2), "minutes")

display(features_df.head())
```

Cell 16 — Standardize the acoustic features

```python
from sklearn.preprocessing import StandardScaler

# Keep only acoustic features
feature_columns = [
    col for col in features_df.columns
    if col != "Uttids"
]

X = features_df[feature_columns]

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original shape:", X.shape)
print("Scaled shape:", X_scaled.shape)

print("\nMean of first 5 features after scaling:")
print(X_scaled[:, :5].mean(axis=0))

print("\nStd of first 5 features after scaling:")
print(X_scaled[:, :5].std(axis=0))
```

Cell 17 — PCA for dimensionality reduction

```python
from sklearn.decomposition import PCA
import numpy as np

# Apply PCA without limiting the number of components
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Explained variance ratio
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print("Original dimensions:", X_scaled.shape[1])
print("PCA dimensions:", X_pca.shape[1])

print("\nExplained variance by first 10 components:")
for i in range(10):
    print(
        f"PC{i+1}: {explained_variance[i]:.4f} "
        f"| Cumulative: {cumulative_variance[i]:.4f}"
    )
```

Cell 18 — Select PCA components

```python
# Keep the first 10 principal components
n_components = 10

X_pca_10 = X_pca[:, :n_components]

print("Original feature dimensions:", X_scaled.shape)
print("Reduced PCA dimensions:", X_pca_10.shape)

print(
    "\nVariance retained:",
    round(cumulative_variance[n_components - 1] * 100, 2),
    "%"
)
```

Cell 19 — K-Means: Test different numbers of clusters

```python
from sklearn.cluster import KMeans
import plotly.graph_objects as go

k_values = range(2, 11)
inertias = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_pca_10)
    inertias.append(kmeans.inertia_)

# Print inertia values
print("K values:", list(k_values))
print("\nInertias:")
for k, inertia in zip(k_values, inertias):
    print(f"K={k}: {inertia:.2f}")

# Interactive dark-theme plot
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=list(k_values),
        y=inertias,
        mode="lines+markers",
        name="Inertia",
        hovertemplate="K = %{x}<br>Inertia = %{y:.2f}<extra></extra>"
    )
)

fig.update_layout(
    title="K-Means Elbow Method",
    xaxis_title="Number of Clusters (K)",
    yaxis_title="Inertia",
    template="plotly_dark",
    hovermode="x unified"
)

fig.show()
```

Cell 20 — K-Means Silhouette Analysis

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

k_values = range(2, 11)
silhouette_scores = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_pca_10)

    score = silhouette_score(
        X_pca_10,
        labels
    )

    silhouette_scores.append(score)

print("Silhouette Scores:")
for k, score in zip(k_values, silhouette_scores):
    print(f"K={k}: {score:.4f}")

# Best K according to silhouette score
best_k = list(k_values)[np.argmax(silhouette_scores)]
best_score = max(silhouette_scores)

print("\nBest K:", best_k)
print("Best Silhouette Score:", round(best_score, 4))
```

Cell 21 — Fit the Final K-Means Model

```python
# Use the best K from silhouette analysis
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

# Fit K-Means and assign cluster labels
kmeans_labels = kmeans.fit_predict(X_pca_10)

# Store cluster labels
features_df["KMeans_Cluster"] = kmeans_labels

print("Number of clusters:", best_k)

print("\nCluster distribution:")
print(features_df["KMeans_Cluster"].value_counts().sort_index())

print("\nCluster centers shape:", kmeans.cluster_centers_.shape)
```

Cell 22 — Compare K-Means Clusters with State Metadata

```python
# Merge cluster labels with metadata
cluster_metadata = features_df[["Uttids", "KMeans_Cluster"]].merge(
    metadata_df[["Uttids", "State"]],
    on="Uttids",
    how="left"
)

# Treat "NA"/missing values consistently
cluster_metadata["State"] = cluster_metadata["State"].fillna("Unknown")

print("State distribution within each K-Means cluster:")
display(
    pd.crosstab(
        cluster_metadata["KMeans_Cluster"],
        cluster_metadata["State"]
    )
)

print("\nState proportions within each cluster:")
display(
    pd.crosstab(
        cluster_metadata["KMeans_Cluster"],
        cluster_metadata["State"],
        normalize="index"
    ).round(3)
)
```

Cell 23 — Evaluate K-Means Clustering

```python
from sklearn.metrics import silhouette_score

# Calculate silhouette score for the final K-Means clustering
kmeans_silhouette = silhouette_score(
    X_pca_10,
    kmeans_labels
)

print("K-Means Silhouette Score:", round(kmeans_silhouette, 4))

# Compare cluster sizes
cluster_counts = features_df["KMeans_Cluster"].value_counts().sort_index()

print("\nCluster sizes:")
for cluster, count in cluster_counts.items():
    percentage = (count / len(features_df)) * 100
    print(f"Cluster {cluster}: {count} samples ({percentage:.1f}%)")
```

Cell 24 — Gaussian Mixture Model: Test Components

```python
from sklearn.mixture import GaussianMixture

components = range(2, 11)

bic_scores = []
aic_scores = []

for n in components:
    gmm = GaussianMixture(
        n_components=n,
        covariance_type="full",
        random_state=42
    )

    gmm.fit(X_pca_10)

    bic_scores.append(gmm.bic(X_pca_10))
    aic_scores.append(gmm.aic(X_pca_10))

print("GMM model comparison:")
print()

for n, bic, aic in zip(components, bic_scores, aic_scores):
    print(f"Components={n}: BIC={bic:.2f}, AIC={aic:.2f}")

# Best number of components
best_gmm_components_bic = components[np.argmin(bic_scores)]
best_gmm_components_aic = components[np.argmin(aic_scores)]

print("\nBest components according to BIC:", best_gmm_components_bic)
print("Best components according to AIC:", best_gmm_components_aic)
```

Cell 25 — Fit the Final Gaussian Mixture Model

```python
# Use the number of components selected by BIC
best_gmm = GaussianMixture(
    n_components=best_gmm_components_bic,
    covariance_type="full",
    random_state=42
)

# Fit GMM
gmm_labels = best_gmm.fit_predict(X_pca_10)

# Store GMM cluster labels
features_df["GMM_Cluster"] = gmm_labels

print("Number of GMM components:", best_gmm_components_bic)

print("\nGMM cluster distribution:")
print(
    features_df["GMM_Cluster"]
    .value_counts()
    .sort_index()
)

print("\nGMM means shape:", best_gmm.means_.shape)
```

Cell 26 — Evaluate GMM Clustering

```python
from sklearn.metrics import silhouette_score

# Calculate silhouette score for GMM clusters
gmm_silhouette = silhouette_score(
    X_pca_10,
    gmm_labels
)

print("GMM Silhouette Score:", round(gmm_silhouette, 4))

# Cluster sizes
gmm_counts = features_df["GMM_Cluster"].value_counts().sort_index()

print("\nGMM cluster sizes:")
for cluster, count in gmm_counts.items():
    percentage = (count / len(features_df)) * 100
    print(f"Cluster {cluster}: {count} samples ({percentage:.1f}%)")
```

Cell 27 — GMM Cluster Membership Probabilities

```python
# Get probability of each sample belonging to each GMM cluster
gmm_probabilities = best_gmm.predict_proba(X_pca_10)

# Maximum probability for each sample
gmm_confidence = np.max(gmm_probabilities, axis=1)

# Store confidence
features_df["GMM_Confidence"] = gmm_confidence

print("GMM probability matrix shape:", gmm_probabilities.shape)

print("\nFirst 10 samples - cluster probabilities:")
print(np.round(gmm_probabilities[:10], 3))

print("\nFirst 10 samples - assigned cluster and confidence:")
display(
    features_df[
        ["Uttids", "GMM_Cluster", "GMM_Confidence"]
    ].head(10)
)

print("\nAverage GMM confidence:",
      round(gmm_confidence.mean(), 4))
```

Cell 28 — Hierarchical Agglomerative Clustering

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

# Perform hierarchical agglomerative clustering
hierarchical = AgglomerativeClustering(
    n_clusters=2,
    linkage="ward"
)

hierarchical_labels = hierarchical.fit_predict(X_pca_10)

# Store cluster labels
features_df["Hierarchical_Cluster"] = hierarchical_labels

# Evaluate clustering
hierarchical_silhouette = silhouette_score(
    X_pca_10,
    hierarchical_labels
)

print("Number of clusters:", 2)
print("Hierarchical Silhouette Score:",
      round(hierarchical_silhouette, 4))

print("\nCluster distribution:")
print(
    features_df["Hierarchical_Cluster"]
    .value_counts()
    .sort_index()
)
```

Cell 34 — Factor Analysis

```python
from sklearn.decomposition import FactorAnalysis

n_factors = 10

factor_analysis = FactorAnalysis(
    n_components=n_factors,
    random_state=42
)

X_factor = factor_analysis.fit_transform(X_scaled)

print("Original feature dimensions:", X_scaled.shape)
print("Factor Analysis dimensions:", X_factor.shape)
```

Cell 35 — Inspect Factor Loadings

```python
factor_loadings = pd.DataFrame(
    factor_analysis.components_.T,
    index=feature_columns,
    columns=[f"Factor_{i+1}" for i in range(n_factors)]
)

print("Factor loading matrix shape:", factor_loadings.shape)

display(
    factor_loadings.round(3)
)
```

Cell 36 — Find the strongest features for each factor

```python
for factor in factor_loadings.columns:
    print(f"\n{factor}")
    print("-" * 40)

    top_features = (
        factor_loadings[factor]
        .abs()
        .sort_values(ascending=False)
        .head(5)
    )

    for feature in top_features.index:
        loading = factor_loadings.loc[feature, factor]
        print(f"{feature}: {loading:.3f}")
```

Cell 37 — Transform the data using the learned factors

```python
factor_columns = [
    f"Factor_{i+1}" for i in range(n_factors)
]

factor_df = pd.DataFrame(
    X_factor,
    columns=factor_columns
)

factor_df["Uttids"] = features_df["Uttids"].values

print("Factor representation shape:", factor_df.shape)

display(factor_df.head())
```

Cell 38 — Cluster the Factor Analysis representation

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

factor_k_values = range(2, 11)
factor_silhouette_scores = []

for k in factor_k_values:
    factor_kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    factor_labels = factor_kmeans.fit_predict(X_factor)

    score = silhouette_score(
        X_factor,
        factor_labels
    )

    factor_silhouette_scores.append(score)

print("Silhouette Scores using Factor Analysis representation:")
print()

for k, score in zip(
    factor_k_values,
    factor_silhouette_scores
):
    print(f"K={k}: {score:.4f}")

best_factor_k = list(factor_k_values)[
    np.argmax(factor_silhouette_scores)
]

best_factor_score = max(
    factor_silhouette_scores
)

print("\nBest K:", best_factor_k)
print("Best Silhouette Score:", round(best_factor_score, 4))
```

Cell 39 — Fit the best Factor Analysis + K-Means model

```python
best_factor_kmeans = KMeans(
    n_clusters=best_factor_k,
    random_state=42,
    n_init=10
)

factor_kmeans_labels = best_factor_kmeans.fit_predict(X_factor)

factor_df["KMeans_Cluster"] = factor_kmeans_labels

print("Best K:", best_factor_k)
print("Silhouette Score:", round(best_factor_score, 4))

print("\nCluster distribution:")
print(
    factor_df["KMeans_Cluster"]
    .value_counts()
    .sort_index()
)
```

Cell 48 — Create the final 1,000-sample state-stratified dataset

```python

final_sample_size = 1000

# Use only the three states with substantial representation
major_states = ["Jharkhand", "MP", "Bihar"]

major_audio_metadata = audio_metadata[
    audio_metadata["State"].isin(major_states)
].copy()

# Proportional sample sizes based on the full archive
state_counts = major_audio_metadata["State"].value_counts()

sample_counts = (
    state_counts / state_counts.sum() * final_sample_size
).round().astype(int)

# Fix rounding so total is exactly 1000
sample_counts.iloc[-1] += (
    final_sample_size - sample_counts.sum()
)

print("Target samples per state:")
print(sample_counts)

# Stratified sampling
final_metadata_parts = []

for state, n in sample_counts.items():
    state_data = major_audio_metadata[
        major_audio_metadata["State"] == state
    ]

    final_metadata_parts.append(
        state_data.sample(
            n=n,
            random_state=42
        )
    )

final_metadata = pd.concat(
    final_metadata_parts,
    ignore_index=True
)

print("\nFinal sample shape:", final_metadata.shape)

print("\nFinal state distribution:")
display(
    final_metadata["State"].value_counts()
)
```

Cell 49 — Extract features from the final 1,000-sample dataset

```python

sample_ids = set(final_metadata["Uttids"])

final_feature_rows = []
start_time = time.time()

with tarfile.open(dataset_path, "r:gz") as tar:
    for member in tar:
        if not member.isfile() or not member.name.endswith(".mp3"):
            continue

        uttid = member.name.split("/")[-1].replace(".mp3", "")

        if uttid not in sample_ids:
            continue

        try:
            audio_bytes = tar.extractfile(member).read()

            y, sr = librosa.load(
                io.BytesIO(audio_bytes),
                sr=16000,
                mono=True
            )

            features = extract_features(y, sr)
            features["Uttids"] = uttid

            final_feature_rows.append(features)

            if len(final_feature_rows) % 200 == 0:
                elapsed = (time.time() - start_time) / 60
                print(
                    f"Processed {len(final_feature_rows)}/1000 "
                    f"| Time: {elapsed:.1f} min"
                )

            if len(final_feature_rows) == 1000:
                break

        except Exception as e:
            print(f"Error processing {uttid}: {e}")

final_features_df = pd.DataFrame(final_feature_rows)

print("\nFinal feature matrix shape:", final_features_df.shape)

print(
    "Time taken:",
    round((time.time() - start_time) / 60, 2),
    "minutes"
)
```

Cell 50 — Merge features with state metadata

```python
final_analysis_df = final_features_df.merge(
    final_metadata[["Uttids", "State"]],
    on="Uttids",
    how="left"
)

print("Final analysis shape:", final_analysis_df.shape)

print("\nState distribution:")
print(final_analysis_df["State"].value_counts(dropna=False))

print("\nMissing states:", final_analysis_df["State"].isna().sum())
```

Cell 51 — Standardize the acoustic features

```python
from sklearn.preprocessing import StandardScaler

# Select only acoustic features
final_feature_columns = [
    col for col in final_analysis_df.columns
    if col not in ["Uttids", "State"]
]

X_final = final_analysis_df[final_feature_columns]

# Standardize
scaler_final = StandardScaler()
X_final_scaled = scaler_final.fit_transform(X_final)

print("Number of acoustic features:", len(final_feature_columns))
print("Scaled feature matrix shape:", X_final_scaled.shape)
```

Cell 52 — PCA and explained variance

```python
from sklearn.decomposition import PCA
import numpy as np

pca_final = PCA()
X_final_pca = pca_final.fit_transform(X_final_scaled)

explained_variance = pca_final.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print("Explained variance by component:")
for i, (var, cum) in enumerate(
    zip(explained_variance, cumulative_variance), start=1
):
    print(
        f"PC{i}: {var:.4f} | "
        f"Cumulative: {cum:.4f}"
    )
```

Cell 53 — Keep 11 PCA components

```python
n_components_final = 11

X_final_pca_reduced = X_final_pca[:, :n_components_final]

print("PCA components selected:", n_components_final)
print("Reduced feature matrix shape:", X_final_pca_reduced.shape)
print(
    "Variance retained:",
    round(cumulative_variance[n_components_final - 1] * 100, 2),
    "%"
)
```

Cell 54 — Find the best number of K-Means clusters

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

silhouette_scores_final = {}

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_final_pca_reduced)

    score = silhouette_score(
        X_final_pca_reduced,
        labels
    )

    silhouette_scores_final[k] = score

    print(f"K={k}: Silhouette Score = {score:.4f}")

best_k_final = max(
    silhouette_scores_final,
    key=silhouette_scores_final.get
)

print("\nBest K:", best_k_final)
print(
    "Best Silhouette Score:",
    round(silhouette_scores_final[best_k_final], 4)
)
```

Cell 55 — Fit the final K-Means model

```python
final_kmeans = KMeans(
    n_clusters=best_k_final,
    random_state=42,
    n_init=10
)

final_clusters = final_kmeans.fit_predict(
    X_final_pca_reduced
)

final_analysis_df["KMeans_Cluster"] = final_clusters

print("Final K-Means model")
print("-------------------")
print("K:", best_k_final)
print(
    "Silhouette Score:",
    round(
        silhouette_score(
            X_final_pca_reduced,
            final_clusters
        ),
        4
    )
)

print("\nCluster sizes:")
print(
    final_analysis_df["KMeans_Cluster"]
    .value_counts()
    .sort_index()
)
```

Cell 56 — Compare clusters with State

```python
cluster_state_counts = pd.crosstab(
    final_analysis_df["KMeans_Cluster"],
    final_analysis_df["State"]
)

cluster_state_percent = pd.crosstab(
    final_analysis_df["KMeans_Cluster"],
    final_analysis_df["State"],
    normalize="index"
) * 100

print("Cluster × State counts:")
print(cluster_state_counts)

print("\n\nCluster × State percentages:")
print(cluster_state_percent.round(2))
```

Cell 57 — Quantify the State–Cluster association

```python
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score

state_labels = final_analysis_df["State"]
cluster_labels = final_analysis_df["KMeans_Cluster"]

ari_score = adjusted_rand_score(
    state_labels,
    cluster_labels
)

nmi_score = normalized_mutual_info_score(
    state_labels,
    cluster_labels
)

print("State–Cluster Association")
print("--------------------------")
print("Adjusted Rand Index (ARI):", round(ari_score, 4))
print("Normalized Mutual Information (NMI):", round(nmi_score, 4))
```
