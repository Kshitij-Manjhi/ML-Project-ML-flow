CELL 32 — Create Audio Feature Extraction Function

```python
def extract_audio_features(audio_object):
    """
    Extract 31 audio features from a LAHAJA audio object.
    """

    # Read audio bytes
    audio_bytes = audio_object["bytes"]
    audio, sr = sf.read(io.BytesIO(audio_bytes))

    # Basic features
    rms = librosa.feature.rms(y=audio)[0]
    zcr = librosa.feature.zero_crossing_rate(audio)[0]

    # Spectral features
    spectral_centroid = librosa.feature.spectral_centroid(
        y=audio, sr=sr
    )[0]

    spectral_bandwidth = librosa.feature.spectral_bandwidth(
        y=audio, sr=sr
    )[0]

    spectral_rolloff = librosa.feature.spectral_rolloff(
        y=audio, sr=sr
    )[0]

    # MFCC
    mfcc = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=13
    )

    # Chroma
    chroma = librosa.feature.chroma_stft(
        y=audio,
        sr=sr
    )

    # Combine features
    features = {
        "duration": len(audio) / sr,
        "rms_energy": np.mean(rms),
        "zero_crossing_rate": np.mean(zcr),
        "spectral_centroid": np.mean(spectral_centroid),
        "spectral_bandwidth": np.mean(spectral_bandwidth),
        "spectral_rolloff": np.mean(spectral_rolloff)
    }

    # Add MFCC features
    for i, value in enumerate(np.mean(mfcc, axis=1), start=1):
        features[f"mfcc_{i}"] = value

    # Add Chroma features
    for i, value in enumerate(np.mean(chroma, axis=1), start=1):
        features[f"chroma_{i}"] = value

    return features


print("Feature extraction function created successfully!")
```

CELL 45 — Bayesian Logistic Regression

```python
import pymc as pm

# Encode the 19 language classes as integers
classes = sorted(y_train.unique())

class_to_int = {cls: i for i, cls in enumerate(classes)}
int_to_class = {i: cls for cls, i in class_to_int.items()}

y_train_encoded = y_train.map(class_to_int).values
y_test_encoded = y_test.map(class_to_int).values

n_features = X_train_scaled.shape[1]
n_classes = len(classes)

with pm.Model() as bayesian_logistic_model:

    # Priors
    weights = pm.Normal(
        "weights",
        mu=0,
        sigma=1,
        shape=(n_features, n_classes)
    )

    bias = pm.Normal(
        "bias",
        mu=0,
        sigma=1,
        shape=n_classes
    )

    # Multinomial logistic regression
    logits = pm.math.dot(X_train_scaled, weights) + bias
    p = pm.math.softmax(logits, axis=1)

    # Likelihood
    observed = pm.Categorical(
        "observed",
        p=p,
        observed=y_train_encoded
    )

    # Variational inference
    approx = pm.fit(
        n=1000,
        method="advi",
        random_seed=42,
        progressbar=True
    )

print("Bayesian Logistic Regression training completed!")
```

CELL 46 — Bayesian Logistic Regression Prediction

```python
# Draw samples from the variational posterior
with bayesian_logistic_model:
    posterior = approx.sample(500)

# Get posterior mean weights and bias
weights_mean = posterior.posterior["weights"].mean(
    dim=("chain", "draw")
).values

bias_mean = posterior.posterior["bias"].mean(
    dim=("chain", "draw")
).values

# Calculate class probabilities for test data
test_logits = np.dot(X_test_scaled, weights_mean) + bias_mean

test_probs = np.exp(
    test_logits - np.max(test_logits, axis=1, keepdims=True)
)

test_probs /= test_probs.sum(axis=1, keepdims=True)

# Predicted class
y_pred_bayesian_encoded = np.argmax(test_probs, axis=1)

y_pred_bayesian = np.array([
    int_to_class[i]
    for i in y_pred_bayesian_encoded
])

accuracy_bayesian = accuracy_score(
    y_test,
    y_pred_bayesian
)

print(f"Bayesian Logistic Regression Accuracy: {accuracy_bayesian:.4f}")
print(f"Accuracy: {accuracy_bayesian * 100:.2f}%")
```
